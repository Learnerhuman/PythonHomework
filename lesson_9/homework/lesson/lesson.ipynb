CSV

```python
import csv
with open("username.csv",'rt') as file:
    reader=csv.reader(file, delimiter=',')
    for row in reader:
        print(row)
```

```python
columns=['username','age']
data=[['a',12],['s',7]]
with open("username.csv",'w', newline='') as file:
    writer=csv.writer(file)
    writer.writerow(columns)
    writer.writerows(data)
```

json

```python
import json
with open("username.json",'r', newline='') as file:
    data=json.load(file)
    print(data)

```
