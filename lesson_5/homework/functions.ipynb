```python
#example
def pos_only(x,/,y):
    return x+y

print(pos_only(3,y=4)) #"x" must be positional


def paas(x,y,/):
    pass
print(paas(x=1,y=2))

#example 2 keyword only

def key_only(*,x,y):
    return x*y

print(key_only(x=3, y=5))# 'x' and 'y' must be key word arguments

```

```python
#arbitary

def sum_all(*args,b):
    return sum(args)

print(sum_all(1,2,3,4,4,b=7))

def print_into(**kwargs):
    print(kwargs)

print_into(name='as',asa='asdd')

def rest(a,b,*args,**kwars):
    pass
print(rest(1,2,23,3,4,a=1,q=1))

```
