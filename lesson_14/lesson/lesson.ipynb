```python
import numpy as np
a1=np.array([1,2,1])
a1.mean()
w=np.array()
(a1*w).sum()/w.sum()#weighted mean
#euclidean norm of vector
np.sqrt(np.sum(a1**2))

#var
((a1-a1.mean())**2).mean()
np.std(a1)**2
#standard deviation
np.std(a1)
#cov

```

Vectorize

```python
@np.vectorize
def categories(x):
    if x<0:
        return "negative"
    elif x==0:
        return 'zero'
    else:
        return 'positive'
categories(a1)
```
