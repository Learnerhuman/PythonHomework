```python
def repeat(n):
    def decorator(func):
        def wrapper(*args,**kwargs):
            for _ in range(n):
                func(*args,**kwargs)
            return wrapper
        return decorator
    
@repeat(4)
def grear(name):
    print(f"Hello, {name}?")
grear("A;lice")
```

```python
def repeat(n):
    def d(func):
        def wrapper(name):
            for _ in range(n):
                func(name)
        return wrapper
    return d

@repeat(6)
def great(name):
    print(f"hello, {name}")

great("yaxshi")

```
