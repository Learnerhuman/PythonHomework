```python
students_values=(
    ('Ali',21),
    ('Garri',23)
)
with sqlite3.connect('students.db') as conn:
    cursor=conn.cursor()
    placeholders=",".join(['?']*3)#so'roqlar ko'p bo'lgan holda. 
    data=cursor.execute(f"Insert into Students values(?,?,?)",students_values)
    data=cursor.executescript(f"Insert into Students values({placeholders})",students_values)
data.executescript #bu ko'p querylarni birdaniga yozayotganda ham kerak,  ; esdan chiqmasin.
```
